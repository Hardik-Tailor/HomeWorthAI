HOUSE PRICE ESTIMATOR

Importing Dataset

```python
import pandas as pd

df = pd.read_excel("Housing_Dataset.xlsx")

```

Data Cleaning 

```python
df["SQUARE_FT"] = df["SQUARE_FT"].round(4)  #putting only 4 values after decimal

#df.to_excel("Housing_Dataset.xlsx", index=False)
```

```python
#making a new column city
df["CITY"] = df["ADDRESS"].str.rsplit(",", n=1).str[-1].str.strip()

#df.to_excel("Housing_Dataset.xlsx", index=False)
```

```python
df.columns = df.columns.str.upper()  #all columns name in capital
#df.to_excel("Housing_Dataset.xlsx", index=False)
```

```python
# Segregating all the cities into tier 1 , tier 2 , and tier 3 

tier1 = [
    "Mumbai","Delhi","Bengaluru","Bangalore","Chennai",
    "Kolkata","Hyderabad","Pune"
]

tier2 = [
    "Ahmedabad","Jaipur","Lucknow","Surat","Kanpur","Nagpur",
    "Indore","Bhopal","Coimbatore","Kochi","Trivandrum",
    "Vizag","Vadodara","Chandigarh","Noida","Gurgaon","Guwahati",
    "Mysore","Amritsar","Patna","Ranchi"
]

def get_tier(city):
    if pd.isna(city):
        return None
    
    city = city.strip().title()

    if city in tier1:
        return 1
    elif city in tier2:
        return 2
    else:
        return 3

df["TIER"] = df["CITY"].apply(get_tier)

df.to_excel("Housing_Dataset.xlsx", index=False)
```
